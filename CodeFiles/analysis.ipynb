```python
my_name = input('Enter my name:')

print(f'My name is:',my_name)

number_of_characters_in_my_name = len(my_name)
print(f'# Characters in My Name:',number_of_characters_in_my_name)

print (f'Spelled out:')
for index, letter in enumerate(my_name):
    print(f'Character #:',index+1,letter)

```
